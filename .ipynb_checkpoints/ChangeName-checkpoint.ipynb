```python
import os
import re
import shutil

input_folder = "masks"      # Folder where current files are
output_folder = "masks_good"    # Folder where renamed files will be saved

os.makedirs(output_folder, exist_ok=True)

# Regex to match: prefix_RVG_###_mask.png
pattern = re.compile(r".*?(RVG_\d+_mask\.png)$")
# Explanation:
# .*?              → any prefix
# (RVG_\d+_mask\.png) → capture RVG_###_mask.png

for filename in os.listdir(input_folder):
    match = pattern.match(filename)
    if match:
        new_name = match.group(1)   # Extract RVG_###_mask.png
        old_path = os.path.join(input_folder, filename)
        new_path = os.path.join(output_folder, new_name)

        shutil.move(old_path, new_path)
        print(f"Moved & renamed: {filename}  -->  {new_name}")
```
